# PySpark Demo

Notebook inicial para validar o ambiente de Engenharia e Ciência de Dados.

```python
from pyspark.sql import SparkSession

spark = (
SparkSession.builder
    .appName('TSE-Analytics-Validation')
    .getOrCreate()
)

print('Spark Version:', spark.version)
```

```python
df = (spark.read
           .format('csv')
           .option('header', 'true')    
           .option('sep', ';')
           .option('encoding', 'ISO-8859-1')
           .load('data/*/consulta_cand_*/*BRASIL.csv'))

df.limit(1000).toPandas()
```

Realizando Agregração com Spark SQL

```python
df.groupBy('ANO_ELEICAO').count().orderBy('ANO_ELEICAO').show()
```

```python
df.printSchema()
```

```python
df.describe().show()
```

```python
output = 'data/demo_parquet'
df.write.mode('overwrite').parquet(output)
print(f'Dados gravados em {output}')
```

```python
df_parquet = spark.read.parquet('data/demo_parquet')
df_parquet.show()
```

```python
spark.stop()
```
